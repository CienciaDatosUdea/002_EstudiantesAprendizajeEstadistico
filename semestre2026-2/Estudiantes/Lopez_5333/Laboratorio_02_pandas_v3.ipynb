````python
#Añadimos la columna DiffPrice  
close_price['DiffPrice'] = abs(close_price['Close'].diff())
close_price['DiffPrice'] = close_price['DiffPrice'].fillna(0)
close_price
````

````python
#Histograma de la columna DiffPrice
import seaborn as sns
import matplotlib.pyplot as plt

sns.histplot(data=close_price, x='DiffPrice', bins=30, zorder=2)
plt.xlabel('Diferencia de Precio (EUR/USD)')
plt.title('Diferencia de Precio (EUR/USD)')
plt.grid(zorder=0)
plt.show()
````

````python
#!pip install fitter
from fitter import Fitter, get_common_distributions, get_distributions
import scipy as scp

print('Distribuciones Posibles para Ajuste de Datos')
display(pd.DataFrame(get_distributions()))

# Ajuste de Distribuciones a la columna DiffPrice   
f = Fitter(close_price['DiffPrice'], distributions=['beta', 'gamma', 'lognorm', 'norm', 'burr', 'expon','weibull_min'], timeout=30)
f.fit()
f.summary()
f_best = f.get_best(method = 'sumsquare_error')

#Extraer los parámetros de la distribución beta ajustada
beta_param = f.fitted_param['beta']


print('--'*50)
print('Parámetros de la distribución beta:')
print(np.array(beta_param).round(4))
print('--'*50)
print('Mejor distribución ajustada:')
print(f_best)
print('--'*50)
````

R:/ La mejor distribucion que se ajusta a los datos de DiffPrice es la distribución exponencial

````python
# Ajuste Mejor Distribución a la columna DiffPrice
sns.histplot(data=close_price, x='DiffPrice', bins=30, stat='density', zorder=2)
x = np.linspace(close_price['DiffPrice'].min(), close_price['DiffPrice'].max(), 1000)

# Ajuste de la distribución exponencial
params = f_best['expon']
y = scp.stats.expon.pdf(x, **params)
plt.plot(x, y, color='red', linewidth=2, label='Distribución Exponencial Ajustada', zorder=3)
plt.xlabel('Diferencia de Precio (EUR/USD)')
plt.title('Ajuste de Distr. Exponencial a la Diferencia de Precio (EUR/USD)')
plt.legend()
plt.grid(zorder=0)
plt.show()
````

**8. Normalizar cada columna respecto a su media y desviación estándar utilizando la fórmula:**

   `(x - mean(x)) / std(x)`

````python
#Normalización Columnas respecto a media y desv. est

norm_cancer_df = cancer_df.copy()

#Excluimos las columnas ID y Diagnosis
for col in norm_cancer_df.columns[2:]:
    mean = norm_cancer_df[col].mean()
    std = norm_cancer_df[col].std()
    norm_cancer_df[col] = (norm_cancer_df[col] - mean) / std

norm_cancer_df
    
````

Usa expresiones regulares para identificar patrones comunes, como `radius1`, `radius2`,`radius3`, etc., y calcular los promedios. Por ejemplo:

```python
re.match(r'^[a-zA-Z_]+', "holamundo12341").group(0)
```

También puedes utilizar el método `startswith()` para buscar estos patrones en los nombres de las columnas.


````python
#Agrupemos Caracteristicas similares 
import re

characteristics = ['RadiusMean', 'TextureMean', 'PerimeterMean',
                   'AreaMean', 'SmoothnessMean', 'CompactnessMean',
                   'ConcavityMean', 'ConcavePointsMean', "SymmetryMean",
                   "FractalDimensionMean"]

for characteristic in characteristics:

    #Obtenemos las columnas que coinciden con la característica
    cols = [col for col in norm_cancer_df.columns
            if re.match(rf'{characteristic}.*', col)]

    if cols:
        #Calculamos la media de las columnas seleccionadas
        norm_cancer_df[characteristic] = norm_cancer_df[cols].mean(axis=1)
        print(f'Característica: {characteristic}, Columnas: {cols}')
````

Usa el rango intercuartílico (IQR) para definir los outliers. Los valores fuera del rango [Q1 - 1.5 * IQR, Q3 + 1.5 * IQR] se consideran outliers. Ejemplo de cómo eliminar estos valores:

```python
Q1 = df['edad'].quantile(0.25)
Q3 = df['edad'].quantile(0.75)
IQR = Q3 - Q1
df = df[~((df['edad'] < (Q1 - 1.5 * IQR)) | (df['edad'] > (Q3 + 1.5 * IQR)))]
```

Otra alternativa es utilizar el puntaje Z para detectar `outliers`:

```python 
df = df[(np.abs(stats.zscore(df['edad'])) < 3)]

**14. ¿Que otro tipo de gráficos pueden ser realizados para entender mejor los datos?**

````python
#Podemos graficar el promedio de las características por diagnóstico usando un gráfico de líneas

vars = norm_cancer_df.columns.difference(['Diagnosis', 'Id'])

promedios = (norm_cancer_df.groupby('Diagnosis')[vars].mean().T)

plt.figure(figsize=(18, 7))
ax = plt.gca()
ax.plot(promedios.index, promedios['B'], marker='o', label='Benigno (B)', color='orange')
ax.plot(promedios.index, promedios['M'], marker='o', label='Maligno (M)', color='blue')

ax.set_xlabel('Características')
ax.set_ylabel('Promedio normalizado')
ax.set_title('Promedio de las características por diagnóstico')
ax.tick_params(axis='x', rotation=90)
ax.grid(alpha=0.3)
ax.legend(title='Diagnosis')

plt.tight_layout()
plt.show()
````
