Ahora vamos a determinar cual es la mejor distribución estadística que se ajusta a la diferencia del precio de cierre cada hora, para ello realizamos lo siguiente:

Determine la diferencia de precio entre horas, agregue una nueva columna llamada DiffPrice, en este punto tu dataframe debe tener solo dos columnas Close, DiffPrice y el indice debe ser el tiempo.

```python
df['DiffPrice'] = df['Close'].shift(-1) - df['Close']
df
```

Determine la mejor distribucion estadística que se ajusta al histograma anterior.

```python
#pip install fitter
```

```python
from fitter import Fitter, get_common_distributions, get_distributions

# Drop NaN values from DiffPrice before fitting
diff_price_clean = df['DiffPrice'].dropna()

f = Fitter(diff_price_clean,
           distributions=['gamma',
                          'lognorm',
                          "beta",
                          "burr",
                          "norm"])
f.fit()
f.summary()
# Indentificamos la mejor distribucion con el error cuadratico medio
f.get_best(method = 'sumsquare_error')
print("La mejor distribución es:",f.get_best(method = 'sumsquare_error'))
```


Normalizar cada columna respecto a su media y desviación estándar.

```python
X=(X - X.mean()) / X.std()
X
```

Agrupar características similares y calcular su promedio. Para las siguientes características:

['RadiusMean', 'TextureMean', 'PerimeterMean', 'AreaMean', 'SmoothnessMean', 'CompactnessMean', 'ConcavityMean', 'ConcavePointsMean', "SymmetryMean", "FractalDimensionMean"]


Ahora, vamos a agrupar las características por su tipo base (por ejemplo, 'Radius', 'Texture') y calcular el promedio de sus tres mediciones correspondientes (`_1`, `_2`, `_3`).

```python
import re

# Lista de los nombres base de las características a agrupar
base_feature_names = [
    'Radius', 'Texture', 'Perimeter', 'Area', 'Smoothness',
    'Compactness', 'Concavity', 'Concavepoints', 'Symmetry', 'Fractaldimension'
]

# DataFrame para almacenar las características agregadas
X_aggregated = pd.DataFrame(index=X.index)

# Calcular el promedio para cada característica base
for base_name in base_feature_names:
    # Seleccionar todas las columnas que comienzan con el nombre base (ej. Radius1, Radius2, Radius3)
    related_cols = [col for col in X.columns if col.startswith(base_name)]
    if related_cols:
        # Calcular la media de estas columnas por fila y añadirla al nuevo DataFrame
        X_aggregated[f'{base_name}Avg'] = X[related_cols].mean(axis=1)

# Mostrar las primeras filas del nuevo DataFrame agregado
display(X_aggregated.head())
```

Crear un gráfico donde se muestre el histograma de la columna RadiusMean separado por las etiquetas B y M, utilizando colores diferentes (naranja y azul, respectivamente) para cada diagnóstico.

```python

merged_df = X_aggregated.join(y['Diagnosis'])

df_b = merged_df[merged_df['Diagnosis'] == 'B']
df_m = merged_df[merged_df['Diagnosis'] == 'M']

plt.figure(figsize=(10, 6))

plt.hist(df_b['RadiusAvg'], bins=30, alpha=0.7, label='Benigno (B)', color='blue', edgecolor='black')
plt.hist(df_m['RadiusAvg'], bins=30, alpha=0.7, label='Maligno (M)', color='orange', edgecolor='black')

plt.title('Histograma de RadiusAvg por Diagnóstico (B vs M)')
plt.xlabel('RadiusAvg Normalizado')
plt.ylabel('Frecuencia')
plt.legend()
plt.grid(True)
plt.show()
```

Generar gráficos de violín para múltiples características. Para las columnas:

RadiusMean
TextureMean
PerimeterMean
AreaMean
SmoothnessMean
CompactnessMean
ConcavityMean
ConcavePointsMean

```python


# List of features to plot
features_to_plot = [
    'RadiusAvg', 'TextureAvg', 'PerimeterAvg', 'AreaAvg', 'SmoothnessAvg',
    'CompactnessAvg', 'ConcavityAvg', 'ConcavepointsAvg', 'Symmetry3', 'Fractaldimension3'
]

# Prepare the DataFrame by combining X_aggregated, X, and y['Diagnosis']
# (This preparation is assumed to be correct from previous steps, ensuring all required columns are present)
plotting_df = X_aggregated.copy()
plotting_df['Symmetry3'] = X['Symmetry3']
plotting_df['Fractaldimension3'] = X['Fractaldimension3']
plotting_df['Diagnosis'] = y['Diagnosis']

# Select only the features to plot and the Diagnosis column
data_for_plot = plotting_df[features_to_plot + ['Diagnosis']]

# Melt the DataFrame to a long format suitable for seaborn.violinplot
data_melted = pd.melt(data_for_plot, id_vars='Diagnosis', var_name='features', value_name='value')

# Generate the violin plot
plt.figure(figsize=(18, 8)) # Increased figure size for better readability of multiple plots
sns.violinplot(x='features', y='value', hue='Diagnosis', data=data_melted, split=True, inner='quart',
               palette={'B': 'blue', 'M': 'orange'})
plt.title('Gráficos de Violín de Características por Diagnóstico (B vs M)')
plt.xlabel('Características')
plt.ylabel('Valor Normalizado')
plt.xticks(rotation=45, ha='right') # Rotate x-axis labels for better readability
plt.legend(title='Diagnóstico')
plt.grid(True, linestyle='--', alpha=0.7)
plt.tight_layout() # Adjust layout to prevent labels from being cut off
plt.show()
```

Determinar y eliminar los valores atípicos (outliers) en la columna RadiusMean.

```python
merged_df.boxplot(column="RadiusAvg", by='Diagnosis', sym='k.', figsize=(18,6))
plt.title('Boxplot de RadiusAvg por Diagnóstico (B vs M)')
plt.suptitle('') # Remove default suptitle to avoid redundancy with custom title
plt.xlabel('Diagnóstico')
plt.ylabel('RadiusAvg Normalizado')
plt.grid(True)
plt.show()
```

```python
Q1 = merged_df['RadiusAvg'].quantile(0.25)
Q3 = merged_df['RadiusAvg'].quantile(0.75)
IQR = Q3 - Q1

merged_df_no_outliers = merged_df[~((merged_df['RadiusAvg'] < (Q1 - 1.5 * IQR)) | (merged_df['RadiusAvg'] > (Q3 + 1.5 * IQR)))]

print(f"Número de filas antes de eliminar outliers: {len(merged_df)}")
print(f"Número de filas después de eliminar outliers: {len(merged_df_no_outliers)}")
```
