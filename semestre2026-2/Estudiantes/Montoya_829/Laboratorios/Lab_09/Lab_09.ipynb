## 4. Estandarización

Calculo medias y desviaciones solo con entrenamiento. Una salida estandarizada negativa indica un nivel inferior a esa media, no un nivel sonoro negativo.

```python
entradas = ['f_Hz', 'alpha_deg', 'c_m', 'U_ms', 'delta_m']

def make_features(frame, kind='original'):
    if kind == 'original':
        return frame[entradas].to_numpy(dtype=float)
    if kind == 'dimensional':
        return np.column_stack([
            np.log10(frame.f_Hz * frame.c_m / frame.U_ms),
            np.deg2rad(frame.alpha_deg),
            np.log10(frame.delta_m / frame.c_m),
            np.log10(frame.U_ms / 1.0)])  # Referencia: 1 m/s
    raise ValueError('Elige original o dimensional')

def prepare(partes, kind='original'):
    X = make_features(df, kind)
    y = df.L_dB.to_numpy()
    it, iv, ie = partes
    sx = StandardScaler().fit(X[it])
    sy = StandardScaler().fit(y[it, None])
    xs = [sx.transform(X[ix]) for ix in partes]
    ys = [sy.transform(y[ix, None]).ravel() for ix in partes]
    return xs, ys, sx, sy

xs, ys, sx, sy = prepare(splits['Aleatoria'])
for etiqueta, X, y in zip(['train', 'valid', 'test'], xs, ys):
    print(etiqueta, 'X:', X.shape, 'y:', y.shape)
print('Medias de X_train escalado:', xs[0].mean(axis=0).round(6))
print('Desviaciones:', xs[0].std(axis=0).round(6))
```

```python
z_120 = sy.transform([[120.]])[0, 0]
recuperado = sy.inverse_transform([[z_120]])[0, 0]
print(f'120 dB se transforma en {z_120:.4f}; al invertir se recuperan {recuperado:.4f} dB.')
assert np.isclose(recuperado, 120.)
```

```python
# Las métricas se expresan en dB después de invertir la transformación del objetivo.
def metrics(y, pred):
    return {'RMSE_dB': np.sqrt(mean_squared_error(y, pred)),
            'MAE_dB': mean_absolute_error(y, pred),
            'R2': r2_score(y, pred)}

def predict_db(model, X, sy):
    pred_z = model.predict(X)
    return sy.inverse_transform(pred_z.reshape(-1, 1)).ravel()

lineal_ejemplo = LinearRegression().fit(xs[0], ys[0])
y_valid = df.iloc[splits['Aleatoria'][1]].L_dB.to_numpy()
pred_valid = predict_db(lineal_ejemplo, xs[1], sy)
print('Referencia lineal en validación:', metrics(y_valid, pred_valid))
```

## 5. Arquitectura y entrenamiento

La red con 5 entradas, dos capas de 32 neuronas y una salida tiene (5×32+32)+(32×32+32)+(32×1+1)=1281 parámetros. Uso ReLU en las capas ocultas y salida lineal.

La tasa de aprendizaje fija el tamaño del paso. max_epochs limita el total de épocas y patience cuenta cuántas pueden pasar sin mejorar validación. Elegiría una época anterior si el error de validación empieza a subir aunque el de entrenamiento siga bajando.

```python
# En cada época medimos train y validación; conservamos los pesos con menor error de validación.
def train_network(xs, ys, hidden, lr, seed=SEED, max_epochs=400, patience=40):
    model = MLPRegressor(hidden_layer_sizes=hidden, activation='relu',
                         solver='adam', learning_rate_init=lr, batch_size=64,
                         alpha=0.0, random_state=seed, shuffle=True,
                         early_stopping=False)
    best_loss = np.inf
    best_model = None
    best_epoch = 0
    espera = 0
    history = []
    for epoch in range(1, max_epochs + 1):
        model.partial_fit(xs[0], ys[0])
        train_loss = mean_squared_error(ys[0], model.predict(xs[0]))
        valid_loss = mean_squared_error(ys[1], model.predict(xs[1]))
        history.append((train_loss, valid_loss))
        if valid_loss < best_loss - 1e-6:
            best_loss = valid_loss
            best_model = copy.deepcopy(model)
            best_epoch = epoch
            espera = 0
        else:
            espera += 1
        if espera >= patience:
            break
    return best_model, np.array(history), best_epoch
```

## 6. Selección de la configuración

Elijo entre las tres configuraciones del enunciado por su RMSE de validación.

```python
configs = [
    {'hidden': (16,), 'lr': 1e-3},
    {'hidden': (32, 32), 'lr': 1e-3},
    {'hidden': (32, 32), 'lr': 3e-3},
]

def select_models(partes, kind='original'):
    xs, ys, sx, sy = prepare(partes, kind)
    y_valid = df.iloc[partes[1]].L_dB.to_numpy()
    linear = LinearRegression().fit(xs[0], ys[0])
    trials, rows = [], []
    for config in configs:
        model, history, best_epoch = train_network(xs, ys, **config)
        score = metrics(y_valid, predict_db(model, xs[1], sy))
        trials.append({'model': model, 'history': history, 'best_epoch': best_epoch})
        rows.append({**config, **score, 'epochs': len(history), 'best_epoch': best_epoch})
    table = pd.DataFrame(rows)
    best = int(table.RMSE_dB.to_numpy().argmin())
    return {'xs': xs, 'sy': sy, 'sx': sx, 'linear': linear,
            'net': trials[best]['model'], 'trials': trials,
            'selection': table, 'best_config': configs[best].copy(),
            'parts': partes, 'kind': kind}

runs = {}
for nombre, partes in splits.items():
    print('Entrenando:', nombre)
    runs[nombre] = select_models(partes)
    display(runs[nombre]['selection'].round(4))
    print('Configuración elegida:', runs[nombre]['best_config'])
```

## 10. Discusión final

El problema consiste en predecir el nivel sonoro escalado de un perfil aerodinámico a partir de cinco entradas: frecuencia, ángulo de ataque, cuerda, velocidad del flujo y espesor de desplazamiento. Las mediciones corresponden a ensayos de un perfil NACA 0012. Una fila es un punto de un espectro bajo determinadas condiciones, no un perfil completo. Por eso varias filas comparten las mismas condiciones físicas y solo cambia la frecuencia.

La primera hipótesis planteó un espectro con un máximo, en vez de una relación monótona entre frecuencia y nivel. Se revisó agrupando por las otras cuatro entradas y ordenando las frecuencias. La segunda propuso una posible asociación positiva entre velocidad y sonido. Aislar esa relación exigiría comparar condiciones equivalentes; las diferencias entre grupos no permiten atribuir todos los cambios únicamente a la velocidad. El máximo del espectro mostrado quedó en 800 Hz, con 129.099 dB.

Se utilizaron dos protocolos. El aleatorio separa aproximadamente 70 % para entrenamiento, 15 % para validación y 15 % para prueba. Evalúa filas de una distribución semejante, pero puede repartir puntos de un mismo espectro entre conjuntos. En el segundo protocolo se reservaron 55.5 m/s para validación y 71.3 m/s para prueba. Este último estudia la transferencia hacia una velocidad superior; también puede cambiar la cobertura de frecuencia, geometría y espesor, así que la diferencia de error no tiene una única explicación.

Las cinco entradas y el objetivo se estandarizaron con las medias y desviaciones del entrenamiento de cada protocolo. Validación y prueba recibieron esas mismas transformaciones. Esto evita que las estadísticas de los conjuntos reservados participen en el ajuste. Las predicciones se devolvieron a la escala original antes de calcular errores. Así, RMSE y MAE están expresados en dB y pueden compararse dentro de cada prueba; las curvas de entrenamiento, en cambio, usan la escala estandarizada de su protocolo.

Se compararon las tres configuraciones del enunciado por RMSE de validación. Ambas particiones eligieron dos capas de 32 neuronas. En la aleatoria, la tasa fue 0.003 y el RMSE de validación 1.617 dB; se recuperó la época 173 de 213. En velocidad reservada, la tasa fue 0.001 y el RMSE 5.281 dB; se recuperó la época 33 de 73. En esta última curva, el MSE de validación subió de 0.584 en la época elegida a 0.798 al terminar. Eso justifica recuperar los pesos anteriores en vez de usar la última época.

En prueba aleatoria, la red obtuvo RMSE 1.561 dB, MAE 1.140 dB y R² 0.945. En velocidad reservada obtuvo RMSE 5.569 dB, MAE 4.226 dB y R² 0.368. La tabla de resultados incluye la media constante y la regresión lineal sobre exactamente las mismas filas. Las diferencias calculadas entre regresión y red indican cuánto mejora o empeora cada métrica, sin cambiar de conjunto durante la comparación. RMSE penaliza más los errores grandes que MAE.

Los gráficos de residuos permiten localizar dispersión y sesgo. En velocidad reservada, la banda inferior a 1000 Hz tiene el mayor MAE: 6.804 dB en 125 observaciones. El panel de errores frente a velocidad concentra todos los puntos en 71.3 m/s, por lo que no permite deducir una pendiente del error respecto a ella. Finalmente, estos resultados describen las condiciones medidas y una separación concreta. No justifican extrapolar a otra geometría de ala ni a cualquier velocidad. Además, el espesor de desplazamiento debe estar disponible para aplicar el modelo. Las métricas reservadas evalúan el procedimiento fijado; cualquier ajuste posterior requeriría una evaluación independiente.

## Conclusión

La red mejora el RMSE aleatorio de 4.675 a 1.561 dB. Al reservar 71.3 m/s, empeora de 4.773 a 5.569 dB frente a la regresión lineal. En esta comparación mejora la interpolación, pero no la extrapolación.
