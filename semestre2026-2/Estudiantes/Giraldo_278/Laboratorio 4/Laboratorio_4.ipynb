## Gradiente descendente

Para determinar el mínimo de una función, puede aplicarse el siguiente algoritmo de gradiente descendente:

- Proponer un número aleatorio inicial $\omega_i$.
- Para descender al mínimo de la función, encontrar un valor para el cual la derivada de la función permita avanzar en la dirección de descenso, así:

\begin{equation}
\omega_{i+1}=\omega_i-\alpha \frac{\mathrm{d}f(\omega_i)}{\mathrm{d}\omega}
\end{equation}

donde $\alpha$ es conocido como la tasa de aprendizaje del algoritmo.

- Evaluar $f(\omega_{i+1})$.
- Iterar hasta encontrar el mínimo de la función.

## Modelo de *machine learning*: solución general

Un modelo general para solucionar un problema de *machine learning* puede ser estructurado como sigue:

### a. Elegir el modelo a emplear

\begin{equation}
h(X,\Theta)
\end{equation}

- En el caso de una regresión lineal, tenemos que $h(X,\Theta=(\theta_0,\theta_1))$:

\begin{equation}
h(X)=\theta_0+\theta_1X
\end{equation}

### b. Elegir la función de coste

- Métrica euclidiana:

\begin{equation}
J(\Theta)=\frac{1}{2m}\sum_{i=1}^{m}\left(h_{\theta}(X^{(i)})-y^{(i)}\right)^2
\end{equation}

- [Lista de funciones de coste que pueden ser empleadas](https://jmlb.github.io/flashcards/2018/04/21/list_cost_functions_fo_neuralnets/)

### c. Aplicar el gradiente descendente iterativamente hasta encontrar el mínimo

\begin{equation}
\Delta \vec{\Theta}=-\alpha \nabla J(\Theta)
\end{equation}

- En el caso de una regresión lineal, tenemos que $h(X,\Theta=(\theta_0,\theta_1))$:

\begin{equation}
\theta_0 := \theta_0-\alpha \frac{\partial J}{\partial \theta_0}
\end{equation}

\begin{equation}
\theta_1 := \theta_1-\alpha \frac{\partial J}{\partial \theta_1}
\end{equation}

Pueden comprobar la solución de la superficie a partir de la forma matricial de la función de coste:

\begin{equation}
J(\Theta)=\frac{1}{2m}\sum_{i=1}^{m}\left(h_{\Theta}(X^{(i)})-y^{(i)}\right)^2
\end{equation}

Sea

\begin{equation}
\Theta^T=[\theta_0,\theta_1]
\end{equation}

y sea la matriz de diseño

\begin{equation}
X=
\begin{bmatrix}
1 & 1 & 1 & \cdots & 1\\
x_1^{(1)} & x_1^{(2)} & x_1^{(3)} & \cdots & x_1^{(m)}
\end{bmatrix}
\end{equation}

de dimensión $(n+1)\times m$. En este caso, como solo se tiene una característica, $n=1$.

Luego, las predicciones del modelo pueden escribirse como

\begin{equation}
\Lambda=\Theta^T X=
\begin{bmatrix}
\theta_0+\theta_1 x_1^{(1)} &
\theta_0+\theta_1 x_1^{(2)} &
\theta_0+\theta_1 x_1^{(3)} &
\cdots &
\theta_0+\theta_1 x_1^{(m)}
\end{bmatrix}
\end{equation}

De esta manera, la función de coste se obtiene comparando estas predicciones con los valores reales $Y$:

\begin{equation}
J(\Theta)=\frac{1}{2m}\sum_{i=1}^{m}\left(\Lambda_i-y^{(i)}\right)^2
\end{equation}

o, de forma equivalente, si definimos

\begin{equation}
Y=
\begin{bmatrix}
y^{(1)} & y^{(2)} & y^{(3)} & \cdots & y^{(m)}
\end{bmatrix},
\end{equation}

entonces

\begin{equation}
J(\Theta)=\frac{1}{2m}\sum_{i=1}^{m}\left((\Theta^T X)_i-Y_i\right)^2
\end{equation}

Por tanto, en Python, la verificación numérica puede realizarse calculando primero las predicciones, luego el error y finalmente el promedio del error cuadrático.

````python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
from sklearn.linear_model import LinearRegression
````

````python
# 1. Generar datos lineales simples (y = 2x + 5)
x = np.arange(1, 11)  # Números del 1 al 10
y = 2 * x + 5  # Relación lineal

df = pd.DataFrame({"X": x, "Y": y})
print(df)
````

\begin{equation}
J(\theta_0,\theta_1)=\frac{1}{2m}\sum_{i=1}^{m}\left(h_{\theta}(x^{(i)})-y^{(i)}\right)^2
\end{equation}

````python
def coste_cuadratico(y_real, y_pred):

    m = len(y_real)
    if m == 0:
        return 0

    # 1 / 2m) * suma((predicciones - reales)^2)
    coste = (1 / (2 * m)) * np.sum((y_pred - y_real) ** 2)
    return coste
````

````python
# Tomamos los datos del dataframe anterior
X_data = df['X'].values
Y_data = df['Y'].values

theta_0_fixed = 0  # se fija theta_0 = 0

# Se barren multiples posibles valores de theta_1
theta_1_values = np.linspace(-10, 10, 100)
cost_values = [] # Aqui ingresan la respuesta de la funcion de costo para los theta_1

# Tomamos los theta_1 que son lo que varian y definimos el y_pred como h(X)=\theta_0+\theta_1X
for theta_1 in theta_1_values:
    y_pred = theta_0_fixed + theta_1 * X_data # Esro muestra el comportamiento lineal del modelo que predicimos, lo que hara el el RSME falle
    cost = coste_cuadratico(Y_data, y_pred) #Tomamos el valor del del dataframe y de la prediccion anterior con los parametos y calculamos el coste cuadratico
    cost_values.append(cost)

# Ahora, para todos los posibles valores de theta_1 encontramos el minimo
min_cost_index = np.argmin(cost_values)

#Esto me da las coordenadas del valor minimo en el espacio coste vs theta_1
best_theta_1 = theta_1_values[min_cost_index]
min_cost = cost_values[min_cost_index]

print(f"Minimum cost: {min_cost:.4f} at theta_1 = {best_theta_1:.4f}")
````

````python
# Plot the cost function
fig1 = plt.figure(figsize=(10, 6))
plt.plot(theta_1_values, cost_values)
plt.scatter(best_theta_1, min_cost, color='red', zorder=5, label=f'Min Cost (θ1={best_theta_1:.2f})')
plt.title('Cost Function vs. theta_1 (with theta_0 = 0)')
plt.xlabel('theta_1')
plt.ylabel('Cost J(theta_0=0, theta_1)')
plt.grid(True)
plt.legend()
plt.show()
````

````python
# Definimos el conjunto de posibles valores de thetas
theta_0_values_range = np.linspace(-10, 10, 100)
theta_1_values_range = np.linspace(-10, 10, 100)

# Se definen ñas poribles configuraciones de los parametros
Thet0, Thet1 = np.meshgrid(theta_0_values_range, theta_1_values_range)

# Una matriz de ceros con la misma dimension de theta
J_values = np.zeros(Thet0.shape)

# Calculate cost for each combination of theta_0 and theta_1

for i in range(Thet0.shape[0]):
    for j in range(Thet0.shape[1]):
        theta_0 = Thet0[i, j]
        theta_1 = Thet1[i, j]
        y_pred = theta_0 + theta_1 * X_data
        J_values[i, j] = coste_cuadratico(Y_data, y_pred)
````

````python
min_idx = np.unravel_index(np.argmin(J_values), J_values.shape)
best_theta_0 = Thet0[min_idx]
best_theta_1 = Thet1[min_idx]
min_cost_3d = J_values[min_idx]
print(f"El mínimo global está en: theta_0 = {best_theta_0:.2f}, theta_1 = {best_theta_1:.2f} con Coste = {min_cost_3d:.4f}")

fig3 = plt.figure(figsize=(12, 8))
ax = fig3.add_subplot(111, projection='3d')
ax.plot_surface(Thet0, Thet1, J_values, cmap='viridis')

ax.set_xlabel('theta_0')
ax.set_ylabel('theta_1')
ax.set_zlabel('Cost J(theta_0, theta_1)')
ax.set_title('Cost Function Surface')
plt.show()
````

Cuando existen variaciones para ambos parametros el punto que minimiza la funcion de coste debe ser la combinacion de prametros que lo minimizan de forma simultanea, evidenciamos un desplazamiento considerable respcto al valor inicial cuando theta_0 era cero y hay un breve aumento en la funcion de coste

# Expresion teorica de la funcion de coste
En el caso de un solo parametro tenemos la expresion:


 \begin{equation}
J(\theta_1)=\frac{1}{2m}\sum_{i=1}^{m}\left(\theta_1x^{(i)})-y^{(i)}\right)^2
\end{equation}

En este caso el minimo de la fucion representa la pendiente anclada al origen con el valor optimo que minimiza la distancia geometrica

Ahora en el caso de dos parametro la funcion de coste es

\begin{equation}
J(\theta_0,\theta_1)=\frac{1}{2m}\sum_{i=1}^{m}\left((\theta_0+\theta_1x^{(i)})-y^{(i)}\right)^2
\end{equation}

En este caso el parametro adicional da un grado extra de libertad sobre la recta cuya pendiente optimiza el problema, esto significa que no esta anclada al origen sino que tambien ttiene libertad sobre sonde ocurre el intercepto

## Gradiente descendente

````python
# Definimos la función y su derivada
def f(x):
    return (x - 4)**2

def df(x):
    return 2 * (x - 4)
````

````python
# 2. Algoritmo de Gradiente Descendente
def gradiente_descendente(x_inicial, alpha, epsilon=1e-4, max_iter=1000):
    x_actual = x_inicial
    historial_x = [x_actual]

    for i in range(max_iter):
        # Calcular el gradiente en el punto actual
        gradiente = df(x_actual)

        # Actualizar el valor de x
        x_siguiente = x_actual - alpha * gradiente
        historial_x.append(x_siguiente)

        # Criterio de parada: el cambio es menor que el error epsilon
        if abs(x_siguiente - x_actual) < epsilon:
            break

        x_actual = x_siguiente

    return x_siguiente, i + 1, historial_x
````

````python
# 3. Pruebas del algoritmo
x_inicial = 0.0  # Propuesta de punto de inicio aleatorio
alphas = [0.1, 0.5, 0.8]
epsilon = 1e-4
````

````python
# Preparar la gráfica
fig, ax = plt.subplots(figsize=(10, 6))
x_vals = np.linspace(-1, 9, 100)
ax.plot(x_vals, f(x_vals), color='black', linewidth=2, label='f(x) = (x-4)²', zorder=1)

# Evaluar para cada tasa de aprendizaje
for alpha in alphas:
    minimo, iteraciones, historial = gradiente_descendente(x_inicial, alpha, epsilon)
    print(f"Alpha: {alpha} | Mínimo en x = {minimo:.5f} | f(x) = {f(minimo):.5f} | Iteraciones: {iteraciones}")

    # Graficar la trayectoria del descenso
    ax.plot(historial, [f(x) for x in historial], marker='o', linestyle='--', alpha=0.7, label=f'Descenso (α={alpha})')

ax.set_title('Gradiente Descendente para f(x) = (x-4)²')
ax.set_xlabel('x')
ax.set_ylabel('f(x)')
ax.legend()
ax.grid(True)
plt.show()
````

````python
# 1. Definir la función F(x, y)
def F(x, y):
    return np.sin(0.5 * x**2 - 0.25 * y**2 + 3) * np.cos(2 * x + 1 - np.exp(y))

#Graficas
resolucion = 100
_X = np.linspace(-3, 3, resolucion)
_Y = np.linspace(-3, 3, resolucion)
_X, _Y = np.meshgrid(_X, _Y)
_Z = F(_X, _Y)

fig = plt.figure(figsize=(15, 6))


ax1 = fig.add_subplot(121, projection='3d')
ax1.plot_surface(_X, _Y, _Z, cmap='coolwarm', alpha=0.8)
ax1.set_title('Superficie 3D de la Función')
ax1.set_xlabel('Eje X')
ax1.set_ylabel('Eje Y')
ax1.set_zlabel('F(x,y)')


ax2 = fig.add_subplot(122)
contour = ax2.contourf(_X, _Y, _Z, 100, cmap='coolwarm')
fig.colorbar(contour, ax=ax2)
ax2.set_title('Mapa de Contorno y Descenso del Gradiente')
ax2.set_xlabel('Eje X')
ax2.set_ylabel('Eje Y')


# Punto de inicio aleatorio
np.random.seed(42)
Theta = np.random.rand(2) * 4 - 2  # Genera coordenadas iniciales

lr = 0.01  # Tasa de aprendizaje
h = 0.001  # incremento
iteraciones = 10000

# Variables para guardar la trayectoria
historial_x = [Theta[0]]
historial_y = [Theta[1]]

for iteracion in range(iteraciones):
    # Derivada Parcial respecto a X
    grad_x = (F(Theta[0] + h, Theta[1]) - F(Theta[0], Theta[1])) / h

    # Derivada Parcial respecto a Y
    grad_y = (F(Theta[0], Theta[1] + h) - F(Theta[0], Theta[1])) / h

    # Actualización de los parámetros
    Theta[0] = Theta[0] - lr * grad_x
    Theta[1] = Theta[1] - lr * grad_y

    # Guardamos cada 10 iteraciones
    if iteracion % 10 == 0:
        historial_x.append(Theta[0])
        historial_y.append(Theta[1])

# Agregamos el último punto
historial_x.append(Theta[0])
historial_y.append(Theta[1])

# Dibujamos la trayectoria en el mapa de contorno
ax2.plot(historial_x, historial_y, color='white', marker='.', markersize=2, alpha=0.5, label='Trayectoria')
# Marcamos el mínimo encontrado con una X roja grande
ax2.plot(Theta[0], Theta[1], color='red', marker='x', markersize=12, markeredgewidth=3, label='Mínimo Encontrado')
ax2.legend()

plt.tight_layout()
plt.show()

# resulra+tados
print(f"Coordenada X óptima: {Theta[0]:.4f}")
print(f"Coordenada Y óptima: {Theta[1]:.4f}")
print(f"Valor mínimo de F(x,y): {F(Theta[0], Theta[1]):.4f}")
````

#Regresion Lineal

````python
# Funcion que pide el problema
np.random.seed(42)
X = np.linspace(0, 1, 100)
y = 0.2 + 0.2*X + 0.02*np.random.random(100)

#Gradiente descediente
m = len(X)
theta_0 = 0.0  # Intercepto inicial
theta_1 = 0.0  # Pendiente inicial
alpha = 0.1    # Tasa de aprendizaje
iteraciones = 5000

for _ in range(iteraciones):
    # Comportamiento lineal
    y_pred = theta_0 + theta_1 * X

    # Cálculo de las derivadas parciales
    d_theta_0 = (1 / m) * np.sum(y_pred - y)
    d_theta_1 = (1 / m) * np.sum((y_pred - y) * X)

    # Actualización de los parametros
    theta_0 -= alpha * d_theta_0
    theta_1 -= alpha * d_theta_1

print(f"Gradiente Descendente")
print(f"Intercepto (theta_0): {theta_0:.4f}")
print(f"Pendiente   (theta_1): {theta_1:.4f}\n")

#Hcacemos lo mismo con la funcion sklearn
X_sklearn = X.reshape(-1, 1)

modelo_sk = LinearRegression()
modelo_sk.fit(X_sklearn, y)

print(f"Scikit-Learn")
print(f"Intercepto: {modelo_sk.intercept_:.4f}")
print(f"Pendiente:  {modelo_sk.coef_[0]:.4f}")

#Graficas

plt.figure(figsize=(10, 6))
plt.scatter(X, y, color='blue', alpha=0.5, label='Datos originales (con ruido)')

#Gradiente Descendente
plt.plot(X, theta_0 + theta_1 * X, color='red', linestyle='--', linewidth=2,
         label=f'Gradiente Descendente (y = {theta_1:.2f}x + {theta_0:.2f})')

#Scikit-Learn
plt.plot(X, modelo_sk.predict(X_sklearn), color='green', linewidth=2, alpha=0.7,
         label=f'Sklearn (y = {modelo_sk.coef_[0]:.2f}x + {modelo_sk.intercept_:.2f})')

plt.title('Comparación de Modelos de Regresión Lineal')
plt.xlabel('X')
plt.ylabel('y')
plt.legend()
plt.grid(True)
plt.show()
````
