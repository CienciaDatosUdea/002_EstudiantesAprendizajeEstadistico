```python

# Create a copy to preserve the original dataset
chete_clean = chete_raw.copy()

# Save original dates
chete_clean["eventDate_original"] = chete_clean["eventDate"]

# Flag records with an imputed date
chete_clean["fecha_imputada"] = False

# Identify the record to correct
mask = (
    (chete_clean["eventID"] == "Chete1-20") &
    (chete_clean["eventDate"].isna())
)

# Assign the date and flag the correction
chete_clean.loc[mask, "eventDate"] = "2023-01-10/2023-01-25"
chete_clean.loc[mask, "fecha_imputada"] = True

# Display the corrected event
display(
    chete_clean.loc[
        chete_clean["eventID"] == "Chete1-20",
        [
            "eventID",
            "individualCount",
            "eventDate_original",
            "eventDate",
            "fecha_imputada"
        ]
    ]
)
```

```python

# Create a copy
data = chete_clean.copy()

# Extract number of sampling nights
data["noches"] = pd.to_numeric(
    data["samplingEffort"].str.extract(
        r"(\d+)\s+noches?"
    )[0],
    errors="coerce"
)

display(data[[
    "eventID",
    "individualCount",
    "samplingEffort",
    "noches"
]].head(10))

```

```python

chete_eventos = (
    data.groupby("eventID", as_index=False)
    .agg(
        mosquitos=("individualCount", "sum"),
        noches_min=("noches", "min"),
        noches_max=("noches", "max")
    )
)

# Identify events with inconsistent sampling effort
chete_eventos["esfuerzo_inconsistente"] = (
    chete_eventos["noches_min"] != chete_eventos["noches_max"]
)

# Assign nights only to consistent events
chete_eventos["noches"] = chete_eventos["noches_max"].where(
    ~chete_eventos["esfuerzo_inconsistente"]
)

display(chete_eventos)
```

```python

# Extract number of nights
chete_clean["noches"] = (
    chete_clean["samplingEffort"]
    .str.extract(r"(\d+)\s+noches?")[0]
    .astype(float)
)

# Group by event
chete_eventos = (
    chete_clean.groupby("eventID", as_index=False)
    .agg(
        eventDate=("eventDate", "first"),
        mosquitos=("individualCount", "sum"),
        noches=("noches", "max")
    )
)

# Calculate mosquitoes per site-night
chete_eventos["mosquitos_por_noche"] = (
    chete_eventos["mosquitos"] / chete_eventos["noches"]
)

display(chete_eventos)
```

```python

# Extract the starting date of each event
chete_eventos["fecha"] = pd.to_datetime(
    chete_eventos["eventDate"].str.split("/").str[0],
    errors="coerce"
)

# Identify the month of each event
chete_eventos["mes"] = (
    chete_eventos["fecha"].dt.to_period("M")
)

# Group by month
chete_mensual = (
    chete_eventos.groupby("mes", as_index=False)
    .agg(
        mosquitos=("mosquitos", "sum"),
        noches=("noches", "sum")
    )
)

# Calculate mosquitoes per site-night
chete_mensual["mosquitos_por_sitio_noche"] = (
    chete_mensual["mosquitos"] / chete_mensual["noches"]
)

# Sort chronologically
chete_mensual = chete_mensual.sort_values("mes")

display(chete_mensual)
```

```python

total_mosquitos = chete_mensual["mosquitos"].sum()
total_noches = chete_mensual["noches"].sum()

promedio = total_mosquitos / total_noches

print("Total de mosquitos:", total_mosquitos)
print("Total de sitio-noches:", total_noches)
print(f"Promedio: {promedio:.2f} mosquitos por sitio-noche")
```

```python


# Create a copy
variacion = chete_mensual.copy()

# Sort chronologically
variacion = variacion.sort_values("mes").set_index("mes")

# Include missing months as NaN (not zero)
todos_los_meses = pd.period_range(
    variacion.index.min(),
    variacion.index.max(),
    freq="M"
)

variacion = variacion.reindex(todos_los_meses)

# Capture rate in the previous month
variacion["tasa_anterior"] = (
    variacion["mosquitos_por_sitio_noche"].shift(1)
)

# Absolute change
variacion["cambio_absoluto"] = (
    variacion["mosquitos_por_sitio_noche"]
    - variacion["tasa_anterior"]
)

# Percentage change
variacion["cambio_porcentual"] = (
    variacion["cambio_absoluto"]
    / variacion["tasa_anterior"] * 100
)

# Percentage change is undefined if previous rate is zero
variacion.loc[
    variacion["tasa_anterior"] == 0,
    "cambio_porcentual"
] = float("nan")

display(variacion[[
    "mosquitos_por_sitio_noche",
    "cambio_absoluto",
    "cambio_porcentual"
]].round(2))
```

```python

# Identify the sampling site
chete_eventos["sitio"] = (
    chete_eventos["eventID"].str.extract(r"^(Chete\d+)")
)

# Group by month and site
por_sitio = (
    chete_eventos.groupby(["mes", "sitio"], as_index=False)
    .agg(
        mosquitos=("mosquitos", "sum"),
        noches=("noches", "sum")
    )
)

# Calculate mosquitoes per night at each site
por_sitio["tasa"] = (
    por_sitio["mosquitos"] / por_sitio["noches"]
)

# Combine both sites and project to 4 nights
chete_fermi = (
    por_sitio.groupby("mes", as_index=False)
    .agg(
        mosquitos_observados=("mosquitos", "sum"),
        mosquitos_por_noche=("tasa", "sum"),
        sitios=("sitio", "nunique")
    )
)

chete_fermi["estimacion_4_noches"] = (
    4 * chete_fermi["mosquitos_por_noche"]
)

# Compare with the original Fermi estimate
chete_fermi["fermi"] = 160

display(chete_fermi)
```

```python

# Extract sampling hours per night from the original records
horas = (
    chete_clean.drop_duplicates("eventID")
    .set_index("eventID")["samplingEffort"]
    .str.extract(r"(\d+)\s*h")[0]
    .astype(int)
)

# Add hours per night to each event
chete_eventos["horas_por_noche"] = (
    chete_eventos["eventID"].map(horas)
)

# Calculate total sampling hours
chete_eventos["horas_totales"] = (
    chete_eventos["noches"] *
    chete_eventos["horas_por_noche"]
)

# Group both sites by month
chete_por_hora = (
    chete_eventos.groupby("mes", as_index=False)
    .agg(
        mosquitos=("mosquitos", "sum"),
        horas_totales=("horas_totales", "sum")
    )
)

# Average mosquitoes per site-hour
chete_por_hora["mosquitos_por_hora"] = (
    chete_por_hora["mosquitos"] /
    chete_por_hora["horas_totales"]
)

display(chete_por_hora)
```

```python
# Partimos de la tabla de eventos ya construida
modelo = chete_eventos.copy()

# Variables conocidas antes de la campaña
modelo["mes_num"] = modelo["fecha"].dt.month
modelo["anio"] = modelo["fecha"].dt.year

# Exposición en sitio-horas
modelo["exposicion"] = (
    modelo["noches"] * modelo["horas_por_noche"]
)

# Variables de estacionalidad
modelo["sin_mes"] = np.sin(
    2 * np.pi * modelo["mes_num"] / 12
)

modelo["cos_mes"] = np.cos(
    2 * np.pi * modelo["mes_num"] / 12
)

# Codificación de los dos sitios
modelo["sitio_2"] = (
    modelo["sitio"] == "Chete2"
).astype(int)

# Variable objetivo: número de mosquitos
columnas = [
    "eventID", "fecha", "sitio",
    "mosquitos", "noches",
    "horas_por_noche", "exposicion",
    "mes_num", "sin_mes", "cos_mes",
    "sitio_2"
]

modelo = modelo[columnas].copy()

# Verificaciones
assert modelo["exposicion"].gt(0).all()
assert modelo["mosquitos"].ge(0).all()

display(modelo.head())
print("Eventos:", len(modelo))
```

```python

# MODELO A: Tasa histórica de captura

# 1. Total de mosquitos capturados
total_mosquitos = modelo["mosquitos"].sum()

# 2. Total de horas de muestreo
total_horas = modelo["exposicion"].sum()

# 3. Estimación de la tasa histórica
tasa_A = total_mosquitos / total_horas

print("Tasa histórica:", round(tasa_A, 3),
      "mosquitos por sitio-hora")

# 4. Predicción para una campaña futura
noches = 4
horas_por_noche = 12
numero_sitios = 2

esfuerzo_futuro = noches * horas_por_noche * numero_sitios

prediccion_A = tasa_A * esfuerzo_futuro

print("Capturas esperadas:", round(prediccion_A))
```

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm
from statsmodels.discrete.discrete_model import NegativeBinomial

# Datos organizados por evento
datos_B = modelo.copy()
datos_B["mes"] = datos_B["fecha"].dt.to_period("M")

meses = sorted(datos_B["mes"].unique())

resultados = []

# Validación temporal: 9 meses iniciales
for i in range(9, len(meses)):

    # Entrenar con campañas anteriores
    train = datos_B[
        datos_B["mes"].isin(meses[:i])
    ]

    # Evaluar en la siguiente campaña observada
    test = datos_B[
        datos_B["mes"] == meses[i]
    ]

    # MODELO A: tasa histórica
    tasa_A = train["mosquitos"].sum() / train["exposicion"].sum()
    pred_A = tasa_A * test["exposicion"].sum()

    # MODELO B: regresión binomial negativa
    X_train = sm.add_constant(
        train[["sitio_2"]], has_constant="add"
    )

    X_test = sm.add_constant(
        test[["sitio_2"]], has_constant="add"
    )

    ajuste = NegativeBinomial(
        train["mosquitos"],
        X_train,
        exposure=train["exposicion"],
        loglike_method="nb2"
    ).fit(disp=False)

    pred_B = ajuste.predict(
        exog=X_test,
        exposure=test["exposicion"]
    ).sum()

    resultados.append({
        "mes": str(meses[i]),
        "observado": test["mosquitos"].sum(),
        "modelo_A": pred_A,
        "modelo_B": pred_B
    })

validacion = pd.DataFrame(resultados)
display(validacion.round(2))
```

```python
for nombre in ["modelo_A", "modelo_B"]:

    error = abs(
        validacion["observado"] - validacion[nombre]
    )

    mae = error.mean()

    wape = (
        error.sum() /
        validacion["observado"].sum()
    ) * 100

    print(nombre)
    print(f"MAE: {mae:.2f} mosquitos")
    print(f"WAPE: {wape:.2f}%")
    print()
```

```python

import pandas as pd
import numpy as np

# 1. Datos mensuales ordenados cronológicamente
datos = (
    chete_por_hora
    .sort_values("mes")
    .reset_index(drop=True)
)

resultados = []

# 2. Validación temporal: comenzamos con 9 meses
for i in range(9, len(datos)):

    # Datos conocidos antes de la campaña
    entrenamiento = datos.iloc[:i]

    # Campaña que vamos a predecir
    prueba = datos.iloc[i]

    # Tasa histórica usando solo datos anteriores
    tasa = (
        entrenamiento["mosquitos"].sum()
        / entrenamiento["horas_totales"].sum()
    )

    # Predicción
    prediccion = tasa * prueba["horas_totales"]

    resultados.append({
        "mes": prueba["mes"],
        "observado": prueba["mosquitos"],
        "predicho": prediccion
    })

# 3. Resultados
validacion_A = pd.DataFrame(resultados)

validacion_A["error_absoluto"] = abs(
    validacion_A["observado"] -
    validacion_A["predicho"]
)

validacion_A["error_porcentual"] = (
    100 * validacion_A["error_absoluto"]
    / validacion_A["observado"]
)

# 4. Métricas de evaluación
MAE = validacion_A["error_absoluto"].mean()

WAPE = (
    validacion_A["error_absoluto"].sum()
    / validacion_A["observado"].sum()
) * 100

print(f"MAE: {MAE:.2f} mosquitos")
print(f"WAPE: {WAPE:.2f}%")

display(validacion_A.round(2))
```

```python

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

datos = (
    chete_por_hora
    .sort_values("mes")
    .reset_index(drop=True)
)

resultados = []

# 9 meses iniciales para entrenamiento
for i in range(9, len(datos)):

    train = datos.iloc[:i]
    test = datos.iloc[i]

    # Estimar tasa con datos de entrenamiento
    tasa = (
        train["mosquitos"].sum()
        / train["horas_totales"].sum()
    )

    # Predicciones dentro del entrenamiento
    pred_train = tasa * train["horas_totales"]

    # Error de entrenamiento
    wape_train = (
        abs(train["mosquitos"] - pred_train).sum()
        / train["mosquitos"].sum()
    ) * 100

    # Predicción del siguiente mes
    pred_test = tasa * test["horas_totales"]

    # Error porcentual del mes de prueba
    error_test = (
        abs(test["mosquitos"] - pred_test)
        / test["mosquitos"]
    ) * 100 if test["mosquitos"] > 0 else np.nan

    resultados.append({
        "mes": str(test["mes"]),
        "observado": test["mosquitos"],
        "predicho": pred_test,
        "wape_train": wape_train,
        "error_test": error_test
    })

validacion_A = pd.DataFrame(resultados)

# WAPE global de prueba
wape_test = (
    abs(validacion_A["observado"]
        - validacion_A["predicho"]).sum()
    / validacion_A["observado"].sum()
) * 100

# Promedio de los errores de entrenamiento
wape_train_promedio = validacion_A["wape_train"].mean()

print(f"WAPE entrenamiento promedio: {wape_train_promedio:.2f}%")
print(f"WAPE test global: {wape_test:.2f}%")

display(validacion_A.round(2))
```

```python

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

# Agrupar los eventos por mes
mensual = (
    chete_eventos.groupby("mes", as_index=False)
    .agg(
        mosquitos=("mosquitos", "sum"),
        noches=("noches", "sum"),
        horas=("horas_totales", "sum")
    )
    .sort_values("mes")
)

# Tasas históricas originales
tasa_hora = mensual["mosquitos"].sum() / mensual["horas"].sum()
tasa_noche = mensual["mosquitos"].sum() / mensual["noches"].sum()

# Bootstrap de bloques mensuales
B = 30000
rng = np.random.default_rng(2026)

datos = mensual[["mosquitos", "noches", "horas"]].to_numpy()

indices = rng.integers(
    0, len(datos),
    size=(B, len(datos))
)

muestras = datos[indices].sum(axis=1)

# Tasas de cada remuestreo
tasas_hora_boot = muestras[:, 0] / muestras[:, 2]
tasas_noche_boot = muestras[:, 0] / muestras[:, 1]

# Intervalos de confianza percentiles del 95 %
ic_hora = np.quantile(tasas_hora_boot, [0.025, 0.975])
ic_noche = np.quantile(tasas_noche_boot, [0.025, 0.975])

print(f"Tasa por hora: {tasa_hora:.2f}")
print(f"IC 95%: {ic_hora.round(2)}")

print(f"\nTasa por noche: {tasa_noche:.2f}")
print(f"IC 95%: {ic_noche.round(2)}")

# Distribución bootstrap de la tasa por hora
plt.figure(figsize=(9, 5))
plt.hist(tasas_hora_boot, bins=45, alpha=0.75)
plt.axvline(tasa_hora, color="black",
            linestyle="--", label="Tasa histórica")
plt.axvline(ic_hora[0], color="red", linestyle=":")
plt.axvline(ic_hora[1], color="red", linestyle=":",
            label="IC 95%")
plt.xlabel("Mosquitos por sitio-hora")
plt.ylabel("Frecuencia bootstrap")
plt.title("Incertidumbre de la tasa histórica - Cheté")
plt.legend()
plt.tight_layout()
plt.show()
```

```python

import numpy as np
import pandas as pd

# Datos históricos agrupados por mes
datos = chete_por_hora.sort_values("mes").copy()

Y = datos["mosquitos"].to_numpy(dtype=float)
E = datos["horas_totales"].to_numpy(dtype=float)

# Campaña futura: 4 noches, 12 horas, 2 sitios
noches = 4
horas_por_noche = 12
sitios = 2

E_futuro = noches * horas_por_noche * sitios

# MODELO A: predicción puntual
tasa_A = Y.sum() / E.sum()
prediccion = tasa_A * E_futuro

# Bootstrap de bloques mensuales
B = 30000
rng = np.random.default_rng(2026)

indices = rng.integers(0, len(Y), size=(B, len(Y)))

Y_boot = Y[indices]
E_boot = E[indices]

# Tasa estimada en cada remuestreo
tasas_boot = Y_boot.sum(axis=1) / E_boot.sum(axis=1)

# Medias esperadas para la futura campaña
mu_futuro = tasas_boot * E_futuro

# Intervalo de confianza para la media
IC95 = np.quantile(mu_futuro, [0.025, 0.975])

# Estimar sobredispersión en cada remuestreo
mu_hist = tasas_boot[:, None] * E_boot

alpha = np.maximum(
    ((Y_boot - mu_hist)**2 - Y_boot).sum(axis=1)
    / (mu_hist**2).sum(axis=1),
    0
)

# Simular las capturas de una campaña futura
capturas = np.empty(B, dtype=int)

nb = alpha > 1e-10

# Binomial negativa cuando existe sobredispersión
r = 1 / alpha[nb]
p = r / (r + mu_futuro[nb])

capturas[nb] = rng.negative_binomial(r, p)

# Poisson cuando no hay sobredispersión estimada
capturas[~nb] = rng.poisson(mu_futuro[~nb])

# Intervalo predictivo
IP95 = np.quantile(capturas, [0.025, 0.975])

print(f"Tasa histórica: {tasa_A:.2f} mosquitos/sitio-hora")
print(f"Predicción puntual: {prediccion:.0f} mosquitos")
print(f"IC 95% de la media: {IC95.round(0)}")
print(f"IP 95% de capturas futuras: {IP95.round(0)}")

print("\nIntervalo predictivo por sitio-hora:")
print(np.round(IP95 / E_futuro, 2))

print("\nIntervalo predictivo por sitio-noche:")
print(np.round(IP95 / (noches * sitios), 2))
```
